In this tournament I made a cooperative agent known as CartersAgent that would work with other agents until betrayed, and would then either undercut them if their defection rate was really high or would essentially randomize whether or not to undercut or invest with a higher chance of investment if the opponent had a history of investment and a higher chance to undercut if the opponent had a history of undercutting, with a slight chance to forgive the opponent. My agent was immediately eliminated in the first round of the bracket tournament, as were most other cooperative agents. It was able to achieve the prestigious rank of third place in the round-robin tournament with a score of 10,804 points, being only 15 points short of the second place winner. There is a clear difference in performance here.

The difference in performance was mostly due to what each tournament was measuring. The bracket tournament focused on seeing if the agents could beat other agents in a 1v1 where the winner was determined with whoever had the most points at the end of the 1v1. This allowed for aggressive agents who often win individual matches while losing more points, to excel. The round-robin tournament focused more on who scored more points overall which let cooperative agents that often gain a lot of points during matches with other cooperative agents, have the upper hand. Between the two tournaments round robin does seem like a better way to test for quality as it lets each agent compete with each other, whereas the bracket tournament can eliminate decent agents too early with bad match ups.
 
These very different tournaments led to two very different champions of each tournament. The adaptive survivor, made by Morgan Smith, was able to win the bracket tournament, being quick to defect and unable to forgive others, letting it often undercut those who thought that she'd forgiven them. The round-robin tournament was won by Stable Cooperator, made by Casey Thomas, which would always cooperate with others, letting him gain a lot of points, even while losing the most rounds. The two tournament champions leaned very heavily into their respective playstyle, with the Round-Robin winner giving up all aggression while the Bracket Champion rejected cooperation very heavily, with no ability to forgive. These two show why my own agent was able to succeed with the round-robin tournament but falter in regards to the bracket tournament, as my agent was very forgiving, making it very cooperative and very similar to Stable Cooperator.
